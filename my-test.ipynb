# OpenAI client with Ollama

This notebook uses the OpenAI Python client to communicate with Ollama running locally.

Before running the Python cells, open PowerShell and run:

```powershell
ollama pull llama3.2
ollama serve
```

If Ollama is already running as a Windows app, `ollama serve` may report that the port is already in use; that is okay.

````python
response = ollama_client.chat.completions.create(
    model=OLLAMA_MODEL,
    messages=[
        {"role": "system", "content": "You are a concise and helpful assistant."},
        {"role": "user", "content": "Explain in two sentences what a local language model is."},
    ],
)

print(response.choices[0].message.content)
````

````python
def ask_ollama(prompt: str) -> str:
    response = ollama_client.chat.completions.create(
        model=OLLAMA_MODEL,
        messages=[{"role": "user", "content": prompt}],
    )
    return response.choices[0].message.content

ask_ollama("Give me three practical ways to learn Python.")
````
